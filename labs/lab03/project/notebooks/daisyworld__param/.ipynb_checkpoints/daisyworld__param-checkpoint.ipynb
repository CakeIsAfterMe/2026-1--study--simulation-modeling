# Daisyworld: базовая визуализация для набора параметров

Повторим базовую визуализацию для нескольких наборов параметров.
Меняются максимальный возраст маргаритки и начальная доля белых
маргариток.

## Активация проекта и загрузка пакетов

```julia
using DrWatson
@quickactivate "project"
using Agents
using DataFrames
using CairoMakie

include(srcdir("daisyworld.jl"))
```

## Параметры эксперимента

Параметры, заданные списком, перебираются: `max_age` принимает
значения 25 и 40, `init_white` --- 0.2 и 0.8. Всего четыре комбинации.

```julia
param_dict = Dict(
    :griddims => (30, 30),
    :max_age => [25, 40],
    :init_white => [0.2, 0.8],
    :init_black => 0.2,
    :albedo_white => 0.75,
    :albedo_black => 0.25,
    :surface_albedo => 0.4,
    :solar_change => 0.005,
    :solar_luminosity => 1.0,
    :scenario => :default,
    :seed => 165,
)
```

Функция `dict_list` создаёт список всех комбинаций:

```julia
params_list = dict_list(param_dict)
length(params_list)
```

## Настройка отображения

```julia
daisycolor(a::Daisy) = a.breed

plotkwargs = (
    agent_color = daisycolor,
    agent_size = 20,
    agent_marker = '✿',
    heatarray = :temperature,
    heatkwargs = (colorrange = (-20, 60),),
)
```

## Запуск для всех наборов

Для каждого набора строим три карты и сохраняем их. Имя файла
строится из параметров функцией `savename`.

```julia
for params in params_list

    model = daisyworld(; params...)

    plt1, _ = abmplot(model; plotkwargs...)

    step!(model, 5)
    plt2, _ = abmplot(model; plotkwargs...)

    step!(model, 40)
    plt3, _ = abmplot(model; plotkwargs...)

    plt1_name = savename("daisyworld", params) * "_step01" * ".png"
    plt2_name = savename("daisyworld", params) * "_step05" * ".png"
    plt3_name = savename("daisyworld", params) * "_step40" * ".png"

    save(plotsdir(plt1_name), plt1)
    save(plotsdir(plt2_name), plt2)
    save(plotsdir(plt3_name), plt3)

    println("Сохранено: ", plt3_name)

end
```
